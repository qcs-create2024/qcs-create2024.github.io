Instead we solve simplified equations by dropping one or the other term in a regime where it can be neglected. For example, when we derive geostrophic balance, we assume that $u, v$ evolve by advection (that is, that the flow speed at a point $\vec x$ changes because faster-or-slower flowing water arrives at that point).
